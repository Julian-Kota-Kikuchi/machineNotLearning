# Correlational Studies
- what are the variables correlated with # of shares
    - the highest correlating variables are not related to polarity..

```python
# 3. Calculate correlations with the target
# We use absolute values to find the strongest relationships (positive or negative)
numeric_df = df.select_dtypes(include=[np.number])
correlations = numeric_df.corr()['shares'].drop('shares').abs()

# 4. Sort and determine the threshold for the top 20%
sorted_corrs = correlations.sort_values(ascending=False)
n_top = int(len(sorted_corrs) * 0.20)

top_20_percent_cols = sorted_corrs.head(n_top)

# Optional: Display the results in a dataframe for a report
top_df = top_20_percent_cols.reset_index()
top_df.columns = ['Variable', 'Absolute Correlation']
```

## Topics and Sentiment
- certain topics should be more polirazied than others

```python
polarity_cols = [col for col in df.columns if 'polarity' in col.lower()]

# Iterate through each category
for category in news_category:
    print(f"--- Statistics for Category: {category} ---")
    
    # Group by the category, select nlp columns, and aggregate
    stats = df.groupby(category)[polarity_cols].agg(['mean', 'median'])
    
    # Display the result
    display(stats)
    print("\n") # Adds a break between tables
```

```python
# difference in mean polarity etc between the category (data_channel_is_lifestyle etc)
all_diffs = []

for category in news_category:
    stats = df.groupby(category)[polarity_cols].mean() 
    delta = stats.loc[1.0] - stats.loc[0.0]
    delta.name = category
    all_diffs.append(delta)

# Combine everything into one summary view
summary_df = pd.DataFrame(all_diffs)
display(summary_df)
```
